```python
import math
```

```python
def direct_g(x):
    try:
        num = math.exp(-x)
        den = 1.0 + math.exp(-x)
        return num / den
    
    except OverflowError:
        return "计算失败：发生 OverflowError (数值溢出)"
    except Exception as e:
        return f"发生未知错误: {e}"
```

```python
def robust_g(x):
    if x >= 0:
        # x 为正时，-x 为负，e^(-x) 极小，不会溢出
        return math.exp(-x) / (1.0 + math.exp(-x))
    else:
        # x 为负时，使用恒等变形公式，此时 x 依然是负数，e^x 极小，完美避开溢出
        return 1.0 / (math.exp(x) + 1.0)
```

```python
#直接计算750
direct_g(750)
```

```python
#稳定算法
robust_g(750)
```

```python
#直接计算-750
direct_g(-750)
```

```python
#稳定算法
robust_g(-750)
```
