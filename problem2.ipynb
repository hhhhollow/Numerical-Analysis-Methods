```python
import math

#原函数
def f(x):
    return x * math.exp(x) - x**2 - 1.0

#一阶导数
def df(x):
    return math.exp(x) * (x + 1.0) - 2.0 * x
```

```python

def newton_method(x0, tol=1e-7, max_iter=50):
    print(f"--- 开始牛顿法迭代 (初值 x0={x0}) ---")
    xk = x0
    for k in range(max_iter):
        fx = f(xk)
        dfx = df(xk)
        
        # 验算：如果导数趋于0，牛顿法失效
        if abs(dfx) < 1e-12:
            print("导数过小，可能发散！")
            return None
            
        x_next = xk - fx / dfx
        
        # 打印每一步的验算过程
        print(f"第 {k+1} 次迭代: x = {x_next:.8f}, 误差(残差) = {abs(f(x_next)):.2e}")
        
        # 满足精度要求则停止
        if abs(x_next - xk) < tol:
            print(f"牛顿法收敛于: {x_next:.8f}\n")
            return x_next
        
        xk = x_next
        
    print("达到最大迭代次数，未收敛！\n")
    
    return xk
```

```python
# 1. 牛顿法 (初值 x0 = 1.0)
root_newton = newton_method(1.0)
```

```python

def secant_method(x0, x1, tol=1e-7, max_iter=50):
    print(f"--- 开始割线法迭代 (初值 x0={x0}, x1={x1}) ---")
    for k in range(max_iter):
        fx0 = f(x0)
        fx1 = f(x1)
        
        # 验算：防止除以0
        if abs(fx1 - fx0) < 1e-12:
            print("函数值差过小，发生除零错误！")
            return None
            
        x_next = x1 - fx1 * (x1 - x0) / (fx1 - fx0)
        
        # 打印每一步的验算过程
        print(f"第 {k+1} 次迭代: x = {x_next:.8f}, 误差(残差) = {abs(f(x_next)):.2e}")
        
        # 满足精度要求则停止
        if abs(x_next - x1) < tol:
            print(f"割线法收敛于: {x_next:.8f}\n")
            return x_next
            
        # 更新节点
        x0 = x1
        x1 = x_next
    print("达到最大迭代次数，未收敛！\n")
    return x1
```

```python
# 2. 割线法 (初值 x0 = 0.0, x1 = 1.0)
root_secant = secant_method(0.0, 1.0)
```

```python
# 最终绝对验算
print(f"牛顿法求得的根代入原方程 f(x) = {f(root_newton):.10e}")
print(f"割线法求得的根代入原方程 f(x) = {f(root_secant):.10e}")
```
