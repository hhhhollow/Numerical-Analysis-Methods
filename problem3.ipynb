```python
#原函数
def f(x):
    return x**5 - 5 * x**2 + 1.0
```

```python
#二分法
def bisection_method(a, b, epsilon):
    
    # 1) 输入连续函数 f(x) 和精度 ε (函数已在外部定义，这里接收区间 a,b 和精度)
    
    # 2) 计算 f(a) 和 f(b). 若 f(a)f(b)>0，则算法失效，停止计算.
    fa = f(a)
    fb = f(b)
    if fa * fb > 0:
        print("算法失效：区间端点函数值同号。")
        return None

    k = 1
    while True:
        # 3) 令 x = (a+b)/2, 并计算 f(x).
        x = (a + b) / 2.0
        fx = f(x)
        
        # 计算当前区间长度，用于停止条件判定
        interval_length = abs(b - a)
        
        #打印
        print(f"第 {k} 次迭代: x = {x:.6f}, |f(x)| = {abs(fx):.2e}, |b-a| = {interval_length:.2e}")
        
        # 若 |f(x)| < ε 或 |b-a| < ε, 停止计算, 输出近似解 x
        if abs(fx) < epsilon or interval_length < epsilon:
            print(f"\n在第 {k} 次输出近似解: x = {x:.8f}")
            return x
            
        # 4) 若 f(a)f(x) < 0, 则令 b=x, 否则令 a=x. 返回第 3 步.
        if fa * fx < 0:
            b = x
        else:
            a = x
            fa = fx  # 工程细节验算：由于 a 更新了，必须同步更新 fa，供下一次循环的 f(a)f(x) < 0 判断使用
            
        k += 1
```

```python
# 求解区间 [-1, 0]，精度 10^-5
root = bisection_method(-1.0, 0.0, 1e-5)
```

```python

print("\n--- 最终计算结果验算 ---")
print(f"求得的根代入原方程 f(x*) = {f(root):.10e}")
```
